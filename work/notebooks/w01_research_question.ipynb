## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

In the starter dataset, there are 30,000 pages across 32 clients. 54.2% of pages are marked as declining, which is more than an editor can review. A simple stale-and-visible rule (180+ days since update and 500+ impressions) flags only 17 pages, so most pages go unranked. In notebook 01, a learned model scored 0.740 on Precision@50 against 0.240 for a hand rule, in one run.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os, subprocess
import pandas as pd

DATA = "data/raw/content_refresh_anonymized.csv"
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.exists(DATA):
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

df = pd.read_csv(DATA)

declining_rate = (df["trend_direction"] == "down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()

print(f"Pages: {len(df):,} across {df['client_id'].nunique()} clients")
print(f"Share of pages marked 'down': {declining_rate:.1%}")
print(f"Pages both stale (180+ days) and visible (500+ impressions): {stale_visible}")

```
