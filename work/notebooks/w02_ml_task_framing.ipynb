## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# --- Tools we need ---
import os, subprocess        # os checks files and folders; subprocess runs a download command
import pandas as pd          # pandas is the tool for working with tables of data (we call it "pd")

# --- Where the data lives ---
DATA = "data/raw/content_refresh_anonymized.csv"                          # the path to the data file
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter" # where to download it from
REPO_DIR = "flyrank-ml-internship-starter"                                # the folder name it downloads into

# --- Download the data if this Colab session doesn't have it yet ---
if not os.path.exists(DATA):                       # if the data file is NOT found...
    if not os.path.isdir(REPO_DIR):                # ...and the folder isn't there either...
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)  # ...download the repo
    os.chdir(REPO_DIR)                             # move into that folder so the file path works

# --- Load the data into a table ---
df = pd.read_csv(DATA)                             # read the CSV file into a table called "df" (one row per page)

# --- Add the target column ---
# Mark a page 1 if its trend is "down" (declining), otherwise 0. This is the proxy label.
df["is_declining_label"] = (df["trend_direction"].str.lower() == "down").astype(int)

# --- Choose which columns to display ---
show = ["client_id", "content_type", "content_age_days", "days_since_last_update",
        "impressions_90d", "avg_position", "ctr", "trend_direction", "is_declining_label"]
show = [c for c in show if c in df.columns]        # keep only the columns that really exist (avoids errors)

# --- Show the results ---
# Print how many pages there are, and what share are marked declining
print(f"Rows (pages): {len(df):,} | Declining rate: {df['is_declining_label'].mean():.1%}")
df[show].head(8)                                   # display the first 8 pages as a table (the last line of a cell shows automatically)
```

Each of the initial rows represents a single page, and pages from the same client can appear multiple times. This is why I plan to validate performance using entire clients. The target variable, is_declining_label, is set to 1 when the trend_direction is “down.”

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

It’s challenging to combine factors like age, impressions, position, and CTR in a single rule, since setting all the cutoffs and weights by hand is both subjective and limiting. Using a model allows these signals to be integrated, providing a ranking for every page without manual thresholds. However, there’s not yet clear evidence that the model outperforms the rule. In notebook 02, both the rule and the tree-based model achieved a Precision@50 of 0.660 using a random split at the page level. That result is based on just one split, and in the test set, the rule only gave non-zero scores to four pages—meaning most of its top 50 were ties at zero. I still need to evaluate performance using a client-level split for a more robust comparison.
